# **Outlier**
An outlier is a data point that is unusually far from the other observations in a dataset.

# **Z-score**

It is a technique which is used to detect outliers.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("data/titanic.csv").drop(columns=["PassengerId", "Ticket", "Cabin", "Name"])
```

```python
df.head()
```

```python
sns.boxplot(data=df, y="Fare")
plt.show()
```

```python
# as fare has a lot of outliers

df["z_fare"] = (df["Fare"]-df["Fare"].mean())/df["Fare"].std()
```

```python
df.head()
```

```python
lower_limit = -3
upper_limit = 3

# these are all oultlier
df[(df["z_fare"]>upper_limit) | (df["z_fare"]<lower_limit)]
```

```python
upper_limit = df["Fare"].mean()+(3*df["Fare"].std())
lower_limit = df["Fare"].mean()-(3*df["Fare"].std())
```

```python
df[(df["z_fare"]>upper_limit) | (df["z_fare"]<lower_limit)]
```
