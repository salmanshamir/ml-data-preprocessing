# **Percentile**

The percentile method identifies extreme values by choosing cutoffs based on the data's distribution.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("data/titanic.csv")
```

```python
df.head()
```

```python
sns.boxplot(df["Fare"])
```

```python
lower_limit = np.percentile(df["Fare"], 1)
upper_limit = np.percentile(df["Fare"], 99)


df[(df["Fare"]<lower_limit) | (df["Fare"]>upper_limit)]
```

```python
print("lower_limit", lower_limit)
print("upper_limit", upper_limit)
```
