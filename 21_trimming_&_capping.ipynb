# **Trimming**

The process of completely deleting data points that fall outside a defined upper or lower limit.


# **Capping**
 The process of replacing outlier values with a specified maximum or minimum threshold rather than deleting them.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("data/titanic.csv").drop(columns=["PassengerId", "Name", "Ticket", "Name", "Cabin"])
```

```python
q1 = np.percentile(df["Fare"], 25)
q3 = np.percentile(df["Fare"], 75)
iqr = q3-q1

lower_limit = q1-(iqr*1.5)
upper_limit = q3+(iqr*1.5)


df[(df["Fare"]<lower_limit) | (df["Fare"]>upper_limit)]
```

```python
# trimming

df[(df["Fare"]>lower_limit) & (df["Fare"]<upper_limit)]
```

```python
# capping 

df["cap_fare"] = df["Fare"].clip(lower_limit, upper_limit)
```
