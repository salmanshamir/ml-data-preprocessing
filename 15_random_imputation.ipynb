# **Random Imputation**

It is also a technique to replace missing value with any random value in the data.
Actually this is not dedicated class for random imputation but we have to implement this by ourself.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("data/titanic.csv").drop(columns=["PassengerId", "Ticket", "Cabin"])
```

```python
df.isna().sum()
```

```python
import warnings

warnings.filterwarnings("ignore")
```

```python
df["Age"][df["Age"].isna()] = df["Age"].dropna().sample(df["Age"].isna().sum()).values
```

```python
df["Embarked"][df["Embarked"].isna()] = df["Embarked"].dropna().sample(df["Embarked"].isna().sum()).values
```
