# **Function Transformer**

A Function Transformer in machine learning is a feature engineering technique used to apply a specific mathematical function to data to modify its distribution

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("data/titanic.csv").drop(columns=["PassengerId", "Cabin", "Ticket"])
```

```python
df.head()
```

```python
# making funciton for applying different function transformer on the dataset


from sklearn.preprocessing import FunctionTransformer


def transformer(funt):
    transform = FunctionTransformer(func=funt)
    Fare_transform = transform.fit_transform(df["Fare"])
    fig, axe = plt.subplots(1, 2, figsize=(12, 4))
    sns.kdeplot(df["Fare"], ax=axe[0])
    axe[0].set_title("Before transformation")
    sns.kdeplot(Fare_transform, ax=axe[1])
    axe[1].set_title("After transformation")
    
    
```

<br>

# **Square Transformer**

```python
transformer(np.square)
```

<br>

# **Square Root Transformer**

```python

transformer(np.sqrt)
```

<br>

# **Reciprocal Transformer**

```python
transformer(np.reciprocal)
```

<br>

# **Logrithmic Transformer**

```python
transformer(np.log1p)
```
