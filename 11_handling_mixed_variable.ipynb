# **Handling mixed variable/column**

Mixed variable are those variable or columns which contains both numeric and alphabetic values.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv("data/titanic.csv", usecols=["Embarked", "Cabin"])

```

```python
df.head()
```

```python
df[df["Embarked"].isna()]
```

```python
# This is just for making mixed variable to practice

df.loc[61, "Embarked"] = "1"
```

```python
df.loc[829, "Embarked"] = "2"
```

```python
df["Cabin"] = df["Cabin"].fillna(df["Cabin"].mode()[0])
```

```python
df.head()
```

```python
# split the cabin column to separate alphabetic and numeric

df["cabin_num"] = df["Cabin"].str.findall(r"\d+").str.join(" ")
```

```python


df["cabin_cat"] = df["Cabin"].str.findall(r'[A-Za-z]+').str.join(" ")
```

```python
df.head()
```

```python
# now it time to separate embarked as it contain sometimes numeric and sometime categorical
df["Embarked"].unique()
```

```python
df["Embarked_num"] = pd.to_numeric(df["Embarked"], errors="coerce", downcast="integer")
```

```python
df["Embarked_cat"] = np.where(df["Embarked_num"].isna(), df["Embarked"], np.nan)
```

```python
temp = df.sample(10)

for i in df["Embarked_num"]:
    if i==1:
        print(i)
        break
```
